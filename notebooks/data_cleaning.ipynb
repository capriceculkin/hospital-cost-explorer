I. Loading the Data

```python
import pandas as pd

HospitalProviderCostReport_Data = pd.read_csv('../data/raw/Hospital_Provider_Cost_Report_2022_Dataset.csv')
```

a. Creating the Data Frame with the Data

```python
HospitalProviderCostReport_DataFrame = pd.DataFrame(HospitalProviderCostReport_Data)
```

II. Cleaning the Data

a: Standardizing Zip Codes

```python
import pandas as pd
import numpy as np

#zip codes
HospitalProviderCostReport_DataFrame['Zip_Code_Clean'] = HospitalProviderCostReport_DataFrame['Zip Code'].astype(str).str[:5]
```

b: Removing Old Zip Code Columns 

```python
HospitalProviderCostReport_DataFrame = HospitalProviderCostReport_DataFrame.drop(
    columns = ['Zip Code'],
    errors = 'ignore'
)
```

```python
HospitalProviderCostReport_DataFrame.to_csv("../data/HospitalProviderCostReport_DataFrame_clean.csv", index=False)
```
